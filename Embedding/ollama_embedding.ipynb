Ollama

Ollama supports embedding models, making it possible to build retrival augmented generation(RAG) applications that combine text prompts with existing documents or other data.

```python
from langchain_community.embeddings import OllamaEmbeddings
#from langchain_ollama import OllamaEmbeddings
```

```python
embedding=(
    OllamaEmbeddings(model="llama3.2:latest") ## By default it use llama2
)
```

```python
embedding
```

```python
r1 =embedding.embed_documents([
    "Alpha is the first letter of Greek alphabet",
    "Beta is the second letter of Greek alphabet",
])
r1
```

```python
len(r1[0])
```

```python
r1[1]
```

```python
# Other Embedding example

embedding=(
    OllamaEmbeddings(model="llama3.2:latest") ## By default it use llama2
)
text = "This is text document"
query_result = embedding.embed_query(text)
query_result
```
