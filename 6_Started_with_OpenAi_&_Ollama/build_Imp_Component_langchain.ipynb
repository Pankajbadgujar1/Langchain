```python
from langchain_core.prompts import ChatPromptTemplate
from langchain_ollama import ChatOllama

# Prompt Template (FREE)  This below prompt gives wrong incorect
# prompt = ChatPromptTemplate.from_messages([
#     ("system", "You are an expert AI Engineer. Provide me answers based on the questions."),
#     ("user", "{input}")
# ])

# prompt = ChatPromptTemplate.from_messages([  # it also give us wrong output
#     ("system",
#      "You are an expert AI Engineer. Answer only if you are confident. "
#      "If you don't know, say 'I am not sure'. Do NOT guess or hallucinate."),
#     ("user", "{input}")
# ])


# This is very strict promt
prompt = ChatPromptTemplate.from_messages([
    ("system",
     "You are an expert AI assistant. "
     "If the question is about a specific product/library/tool and you are not 100% sure, say: "
     "'I am not sure' and ask the user for clarification. "
     "Never invent facts."),
    ("user", "{input}")
])


# Ollama Model (FREE - Local)
llm = ChatOllama(model="llama3.2:latest")

# Combine prompt + model
chain = prompt | llm

# Run
response = chain.invoke({"input": "Can you tell me about LangSmith in LangChain?"})
print(response.content)




```

```python
## stroutput Parser

from langchain_core.output_parsers import StrOutputParser
output_parser=StrOutputParser()
chain=prompt|llm|output_parser

response=chain.invoke({"input":"Can you tell me about python programming langauge"})
print(response)
```
